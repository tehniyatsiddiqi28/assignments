```python
import pandas as pd 
```

```python
df = pd.read_csv(r"C:\Users\hp\Downloads\teaching_prices_cleaned.csv")
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df["item"] = df["item"].str.strip().str.title()
```

```python
df = df.sort_values("date")
```

```python
stats = df.groupby("item")["price"].agg(
    mean="mean",
    median="median",
    first="first",
    last="last"
)
stats
```

```python
stats["pct_change"] = (
    (stats["last"] - stats["first"]) / stats["first"]
) * 100
stats
```

```python
stats.round(2)
stats
```
